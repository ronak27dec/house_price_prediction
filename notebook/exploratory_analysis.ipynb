```python
import csv
from dataclasses import dataclass

@dataclass
class Model:
    area_sqft: str
    rooms: float
    build_year: int
    location: str
    street_type: str
    furnishing: str
    property_type: str
    has_pool: bool
    price: float


data_model = []

with open('../data/raw/dataset_2.csv', mode='r') as f:
    reader = csv.DictReader(f)

    for row in reader:
        obj = Model(area_sqft=row['Area_SqFt'],
                    rooms=row['Rooms'],
                    build_year=row['Build_Year'],
                    location=row['Location'],
                    street_type=row['Street_Type'],
                    furnishing=row['Furnishing'],
                    property_type=row['Property_Type'],
                    has_pool=row['Has_Pool'],
                    price=row['Price'])
        data_model.append(obj)


    data_model

```
